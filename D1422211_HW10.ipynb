第一題

```python
def ad_report(impressions, clicks, target=3.0):
    if impressions == 0:
        return (None, False)
    ctr = clicks / impressions * 100
    passed = ctr >= target
    return (ctr, passed)


student_id = input("請輸入完整學號：").strip().upper()
tail = student_id[-3:]
a = int(student_id[-3])
b = int(student_id[-2])
c = int(student_id[-1])
s = a + b + c

ads = [
    ("搜尋廣告", 1000 + 100 * a, 30 + 5 * b),
    ("社群廣告", 1200 + 100 * b, 24 + 4 * c),
    ("影音廣告", 0, 0),
]
target = 2.5 + 0.1 * c

print(f"學號：{student_id}")
print(f"CTR門檻：{target:.2f}%")

passed_count = 0
for name, impressions, clicks in ads:
    ctr, passed = ad_report(impressions, clicks, target=target)
    if passed:
        passed_count += 1
    ctr_text = "無曝光資料" if ctr is None else f"{ctr:.2f}%"
    print(f"{name}：曝光={impressions}，點擊={clicks}，CTR={ctr_text}，達標={passed}")

print(f"達標廣告數：{passed_count}")
print(f"預設門檻測試：{ad_report(1000, 30)}")
print(f"零點擊測試：{ad_report(1000, 0)}")
```

```python
def collect_codes(raw_codes, saved=None):
    result = list(saved) if saved is not None else []
    for item in raw_codes:
        code = item.strip().upper()
        if code and code not in result:
            result.append(code)
    return result


student_id = input("請輸入完整學號：").strip().upper()
tail = student_id[-3:]
a = int(student_id[-3])
b = int(student_id[-2])
c = int(student_id[-1])
s = a + b + c

saved = ["BASE"]
raw_a = [" welcome" + tail + " ", "vip" + tail, "", "WELCOME" + tail, "   "]
raw_b = [" new" + tail + " ", " welcome" + tail + " "]

activity_a = collect_codes(raw_a, saved=saved)
activity_b = collect_codes(raw_b)

print(f"學號：{student_id}")
print(f"活動A：{activity_a}")
print(f"活動B：{activity_b}")
print(f"原始saved：{saved}")
print(f"原始raw_a：{raw_a}")
print(f"空資料：{collect_codes([])}")
```

第三題

```python
def cumulative_spend(daily_spend, n):
    if n == 0:
        return 0
    return cumulative_spend(daily_spend, n - 1) + daily_spend[n - 1]


student_id = input("請輸入完整學號：").strip().upper()
tail = student_id[-3:]
a = int(student_id[-3])
b = int(student_id[-2])
c = int(student_id[-1])
s = a + b + c

daily_spend = [100 + 10 * a, 150 + 10 * b, 0, 200 + 10 * c, 50 + 10 * s]
n = 3 + c % 3

print(f"學號：{student_id}")
print(f"每日費用：{daily_spend}")
print(f"指定n：{n}")
print(f"指定前{n}天累計：{cumulative_spend(daily_spend, n)}元")

for day in range(6):
    print(f"前{day}天：{cumulative_spend(daily_spend, day)}元")

print(f"空資料：{cumulative_spend([], 0)}元")
```

第四題

```python
def select_clicks(raw_clicks, min_clicks=10):
    stripped = map(lambda s: s.strip(), raw_clicks)
    non_empty = filter(None, stripped)
    numbers = list(map(int, non_empty))
    qualified = list(filter(lambda x: x >= min_clicks, numbers))
    return (numbers, qualified)


student_id = input("請輸入完整學號：").strip().upper()
tail = student_id[-3:]
a = int(student_id[-3])
b = int(student_id[-2])
c = int(student_id[-1])
s = a + b + c

raw_clicks = [
    str(10 + a),
    "0",
    "",
    " " + str(5 + b) + " ",
    "   ",
    str(15 + c),
    str(5 + b),
]
min_clicks = 5 + b

numbers, qualified = select_clicks(raw_clicks, min_clicks=min_clicks)

print(f"學號：{student_id}")
print(f"點擊門檻：{min_clicks}")
print(f"清理結果：{numbers}")
print(f"合格結果：{qualified}")
print(f"合格筆數：{len(qualified)}")
print(f"合格合計：{sum(qualified)}")

for rank, count in enumerate(sorted(qualified, reverse=True), start=1):
    print(f"第{rank}名：{count}次")

_, qualified_100 = select_clicks(raw_clicks, min_clicks=100)
print(f"門檻100合格結果：{qualified_100}")
print(f"門檻100合格筆數：{len(qualified_100)}")
print(f"門檻100合格合計：{sum(qualified_100)}")
if not qualified_100:
    print("無符合門檻的素材")
```
